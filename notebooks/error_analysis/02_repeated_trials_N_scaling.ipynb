```python
from pathlib import Path
import os
import sys
import json
import numpy as np
import pandas as pd


def find_project_root(start: Path) -> Path:
    current = start.resolve()
    while current != current.parent:
        if (current / "src").exists():
            return current
        current = current.parent
    raise RuntimeError("Could not locate project root containing src/")


PROJECT_ROOT = find_project_root(Path.cwd())
os.chdir(PROJECT_ROOT)

SRC = PROJECT_ROOT / "src"
ARTIFACTS = PROJECT_ROOT / "artifacts"

if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

print("PROJECT_ROOT:", PROJECT_ROOT)
print("CWD:", Path.cwd())
print("ARTIFACTS:", ARTIFACTS)
```

# Levi

```python
SYSTEM_NAME = "levi"
PROFILE_NAME = "levi_velocity_grid"
INTERACTION_TYPE = "alignment"
SYSTEM_ORDER = 2

FIT_END_TIME = 0.75
NUM_TRIALS = 10

TRAIN_FIT_SPLIT = "train_fit"
TRAIN_FULL_SPLIT = "train"
TEST_SPLIT = "test"
TEST_4N_SPLIT = "test_4N"

TRAIN_FIT_NAME_TEMPLATE = "dataset_train_fit_trial_{trial:03d}"
TRAIN_FULL_NAME_TEMPLATE = "dataset_train_trial_{trial:03d}"
TEST_DATASET_NAME = "dataset_test"
TEST_4N_DATASET_NAME = "dataset_test_4N"

function_value_dir = (
    ARTIFACTS
    / SYSTEM_NAME
    / "evaluations"
    / "repeated_trials"
    / "function_values"
)
```

```python
from ips_learning.analysis.scalability_evaluation import ScalabilityEvaluationRunner
from ips_learning.analysis.scalability_trial_runner import ScalabilityTrialConfig

config = ScalabilityTrialConfig(
    system_name=SYSTEM_NAME,
    profile_name=PROFILE_NAME,
    interaction_type=INTERACTION_TYPE,
    system_order=SYSTEM_ORDER,

    num_trials=NUM_TRIALS,
    train_horizon_end=FIT_END_TIME,

    train_fit_split=TRAIN_FIT_SPLIT,
    train_full_split=TRAIN_FULL_SPLIT,
    test_split=TEST_SPLIT,
    test_4N_split=TEST_4N_SPLIT,

    train_fit_name_template=TRAIN_FIT_NAME_TEMPLATE,
    train_full_name_template=TRAIN_FULL_NAME_TEMPLATE,
    test_dataset_name=TEST_DATASET_NAME,
    test_4N_dataset_name=TEST_4N_DATASET_NAME,

    radial_basis_family="bspline",
    radial_degree=1,
    radial_n_basis=8,

    environment_basis_family="bspline",
    environment_degree=1,
    environment_n_basis_per_dim=10,

    solver_method="lstsq",
    ridge_lambda=1e-8,
    svd_tol=1e-12,

    rollout_method="Radau",
    rollout_rtol=1e-6,
    rollout_atol=1e-6,
    save_rollouts=True,

    save_function_values=True,
    function_value_dir= function_value_dir,
)

levi_outputs = ScalabilityEvaluationRunner(
    SYSTEM_NAME=SYSTEM_NAME,
    PROFILE_NAME=PROFILE_NAME,
    INTERACTION_TYPE=INTERACTION_TYPE,
    FIT_END_TIME=FIT_END_TIME,
    NUM_TRIALS=NUM_TRIALS,
    ARTIFACTS=ARTIFACTS,
    config=config,
).outputs
```

```python
for key, value in levi_outputs.items():
    print(f"\n--- {key} ---")

    if isinstance(value, pd.DataFrame):
        print(value.head())
    else:
        print(value)
```
